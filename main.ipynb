```python
#Importación de librerías
import sys
import os
import numpy as np
import matplotlib.pyplot as plt

from lectura import leer_imagen, mascara_grises, dir_imagenes
```

```python
#Cargar lista de imágenes desde el directorio
img_dir = r"C:\Users\delL-\Documents\UPIIT-IPN\Visión Artificial\VA_Proyecto-Final\Datasets\train_images_labels_targets\train\images"

imagenes = dir_imagenes(img_dir)
print("Total imágenes encontradas:", len(imagenes))

img = leer_imagen(imagenes[0])
print("Shape:", img.shape)
```

```python
#Comparar las imagenes entre pre y post desastre
img_post = leer_imagen(imagenes[0])

ruta_pre = imagenes[0].replace("_post_disaster","_pre_disaster")
img_pre = leer_imagen(ruta_pre)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)  
plt.imshow(img_pre)
plt.title("Pre-desastre")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(img_post)  
plt.title("Post-desastre")
plt.axis('off')

plt.show()
```

```python
mask_dir = r"C:\Users\delL-\Documents\UPIIT-IPN\Visión Artificial\VA_Proyecto-Final\Datasets\train_images_labels_targets\train\targets"

nombre = os.path.basename(imagenes[0]).replace("_post_disaster.png","_target.png")
ruta_mask = os.path.join(mask_dir, nombre)

mask = mascara_grises(ruta_mask)

plt.figure(figsize=(5,5))
plt.imshow(mask, cmap='gray')
plt.title("Máscara de daño")
plt.axis('off')         
plt.show()

print("Valores únicos en la máscara:", np.unique(mask)) 

```
